## Ames Housing Exploratory Data Analysis (EDA)

This repository contains a comprehensive, academic-grade Exploratory Data Analysis (EDA) on the Ames Housing Dataset (train.csv). The project investigates the structural, categorical, and geographic drivers of home prices across 1,460 observations and 81 features.

#### Analytical Workflow & Notebook Structure

The analysis in Src/ames_housing_eda.ipynb is structured into five core diagnostic phases:

**1. Structure & Dimension Audit**

Loads train.csv relative to the Src/ folder using Python's os.path.join("..", "Data", "train.csv").

Confirms dataset dimensions of 1,460 rows × 81 columns (df_train.shape).

Audits data types via df_train.info(), identifying 38 numerical features (int64, float64) and 43 categorical/string attributes.

Generates primary summary statistics using df_train.describe().T.

**2. Missing Values Analysis**

Quantifies null counts using df_train.isnull().sum().

Identifies features with extreme structural missingness: PoolQC (99.52%), MiscFeature (96.30%), Alley (93.77%), and Fence (80.75%).

Renders a custom palette bar plot (Purples_r) saved to ../Results/missing_values_top.png.

Concludes that high missingness reflects the physical absence of features (e.g., no pool, no alley) and recommends "None" categorical encoding.


**3. Univariate Analysis: Numerical Variables**

Evaluates non-normality and skewness using 1x2 subplots with Kernel Density Estimation (sns.histplot with kde=True):

SalePrice: Right-skewed distribution ranging from $34,900 to $755,000 (Median: $163,000, Mean: $180,921).

GrLivArea: Right-skewed distribution ranging from 334 to 5,642 sq ft (Median: 1,464 sq ft).

Generates a categorical frequency count plot for ordinal material quality (OverallQual), highlighting modal clustering around ranks 5 (~395 homes) and 6 (~375 homes).


**4. Bivariate Analysis & Outlier Detection**

Filters numerical features (df_train.select_dtypes(include=['float64', 'int64'])) and calculates the Pearson correlation matrix (.corr()).

Isolates top predictors with $\vert{}r\vert{} > 0.5$ against SalePrice in a coolwarm heatmap (../Results/correlation_matrix.png):

- OverallQual ($r = 0.79$): Exponential valuation growth from rank 5 upwards.
- GrLivArea ($r = 0.71$): Strong linear relationship, revealing 2 influential partial sale outliers (>4,500 sq ft selling under $200,000).
- GarageCars ($r = 0.64$): Peak median pricing at 3-car capacity ($300,000) with a non-linear drop at 4 cars.
- TotRmsAbvGrd: Direct structural correlation with total living area.

**5. Categorical Variable Impact**

Evaluates geographic price distribution across 25 neighborhoods using sns.boxplot (Set3 palette).

Reveals a significant price gap between low-median areas (IDOTRR, MeadowV < $100,000) and luxury districts (NridgHt, NoRidge, StoneBr > $300,000).
